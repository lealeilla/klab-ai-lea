# Machine Learning Training

In this section, I will use the cleaned heart disease dataset to build and compare two machine learning models: Linear Regression and Random Forest.

```python
import pandas as pd

# Load the cleaned dataset
df = pd.read_csv("../data/processed/heart_disease_cleaned.csv")

# Display the first five rows
df.head()
```

```python
#Check the cleaned dataset
print("Dataset shape:", df.shape)

print("\nColumns:")
print(df.columns.tolist())

print("\nMissing values:")
print(df.isna().sum())

print("\nData types:")
print(df.dtypes)
```

```python
# Convert Heart Disease Status into a numerical target

df["Heart_Disease_Target"] = df["Heart Disease Status"].map({
    "No": 0,
    "Yes": 1
})

print(df[["Heart Disease Status", "Heart_Disease_Target"]].head())
```

```python
# Select the same features for both models

features = [
    "Age",
    "Blood Pressure",
    "Cholesterol Level",
    "BMI",
    "Sleep Hours"
]

X = df[features]
y = df["Heart_Disease_Target"]

print("Features (X):")
print(X.head())

print("\nTarget (y):")
print(y.head())
```

## Train-Test Split

I split the dataset into 80% training data and 20% testing data. The training data will be used to train the models, while the testing data will be used to evaluate their performance. I use `stratify=y` to maintain the same proportion of heart disease cases in both datasets.

```python
from sklearn.model_selection import train_test_split

# Split the data into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training features:", X_train.shape)
print("Testing features:", X_test.shape)
print("Training target:", y_train.shape)
print("Testing target:", y_test.shape)

print("Training target distribution:")
print(y_train.value_counts())

print("\nTesting target distribution:")
print(y_test.value_counts())
```

```python
print("Training features:", X_train.shape)
print("Testing features:", X_test.shape)

print("\nTraining target distribution:")
print(y_train.value_counts())

print("\nTesting target distribution:")
print(y_test.value_counts())
```

## Linear Regression

Linear Regression is used as the first model and baseline for comparison. The model learns the relationship between the five selected health features and the numerical heart disease target (0 = No, 1 = Yes).

```python
from sklearn.linear_model import LinearRegression

# Create the Linear Regression model
linear_model = LinearRegression()

# Train the Linear Regression model
linear_model.fit(X_train, y_train)

# Make predictions on the test data
linear_predictions = linear_model.predict(X_test)

print("First 10 Linear Regression predictions:")
print(linear_predictions[:10])
```

```python
# Convert predictions into Yes/No  
#logic is Prediction < 0.5  → 0 → No or Prediction ≥ 0.5  → 1 → Yes

linear_classes = (linear_predictions >= 0.5).astype(int)

print("First 10 predicted classes:")
print(linear_classes[:10])
```

## Evaluate Linear Regression

```python
from sklearn.metrics import accuracy_score, mean_squared_error

linear_accuracy = accuracy_score(
    y_test,
    linear_classes
)

linear_mse = mean_squared_error(
    y_test,
    linear_predictions
)

print("Linear Regression Accuracy:", round(linear_accuracy, 4))
print("Linear Regression MSE:", round(linear_mse, 4))
```

```python
comparison = pd.DataFrame({
    "Actual": y_test.values,
    "Linear_Prediction": linear_predictions,
    "Predicted_Class": linear_classes
})

comparison.head(10)
```

## Random Forest

Random Forest is the second model. It uses multiple decision trees to learn patterns between the selected health features and heart disease status. The model uses the same features and train-test split as Linear Regression so that the comparison is fair.

```python
from sklearn.ensemble import RandomForestClassifier

# Create the Random Forest model
random_forest = RandomForestClassifier(
    n_estimators=100,
    random_state=42
)

# Train the Random Forest model
random_forest.fit(X_train, y_train)


rf_predictions = random_forest.predict(X_test)

rf_accuracy = accuracy_score(y_test, rf_predictions)

print("Random Forest Accuracy:", round(rf_accuracy, 4))
```

## Model Comparison

The Linear Regression model achieved an accuracy of 80.00%, while the Random Forest model achieved an accuracy of 80.05%.

Random Forest performed slightly better based on accuracy, but the difference was only 0.05 percentage points. Therefore, the two models performed almost the same on this dataset and with the selected features.

## Confusion Matrix

```python
from sklearn.metrics import confusion_matrix, classification_report

print("Linear Regression Confusion Matrix:")
print(confusion_matrix(y_test, linear_classes))

print("\nLinear Regression Classification Report:")
print(classification_report(y_test, linear_classes))
```

```python
print("Random Forest Confusion Matrix:")
print(confusion_matrix(y_test, rf_predictions))

print("\nRandom Forest Classification Report:")
print(classification_report(y_test, rf_predictions))
```

Random Forest achieved slightly higher accuracy (80.05%) than Linear Regression (80.00%). However, the difference is very small, so neither model can be considered clearly superior based on accuracy alone. Because the target classes are imbalanced, confusion matrix and classification metrics should also be considered when evaluating the models.
